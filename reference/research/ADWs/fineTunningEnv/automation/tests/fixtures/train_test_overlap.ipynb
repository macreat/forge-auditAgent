# Train/test overlap
The test split reuses training samples. Reproduces the TRAIN ∩ TEST ≠ ∅ failure (severity 9, category `splits`).

```python
import numpy as np

# 80 training samples (indices 0..79).
train_ids = np.arange(0, 80)

# BUG: test reuses samples 40..119, overlapping training on 40..79.
test_ids = np.arange(40, 120)
```
