# Checkpoint not restored
The best checkpoint is saved but never loaded back before final evaluation. Reproduces the missing-checkpoint-restoration failure (severity 9, category `checkpoints`).

```python
import pickle


class SimpleModel:
    def __init__(self):
        self.weight = [0.5, -0.25, 0.75]

    def state_dict(self):
        return {"weight": list(self.weight)}


model = SimpleModel()
```

```python
# The "best" checkpoint is saved to disk...
with open("best.pt", "wb") as fh:
    pickle.dump(model.state_dict(), fh)
```

```python
# BUG: final evaluation uses the in-memory model, never restoring best.pt.
accuracy = 0.92
```
