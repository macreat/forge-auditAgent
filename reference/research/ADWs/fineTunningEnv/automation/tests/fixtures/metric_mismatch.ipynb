# Metric mismatch
Accuracy is computed with a wrong denominator so it exceeds 1. Reproduces the metric-out-of-range failure (severity 8, category `metrics`).

```python
import numpy as np

y_true = np.array([0, 1, 0, 1])
y_pred = np.array([0, 1, 1, 1])

# BUG: 3 correct out of sum(y_true)=2 -> accuracy = 1.5, outside [0, 1].
accuracy = float(np.sum(y_pred == y_true) / np.sum(y_true))
```
