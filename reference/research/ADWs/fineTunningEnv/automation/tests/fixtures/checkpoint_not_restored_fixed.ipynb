# Checkpoint restored (fixed)
The best checkpoint is now loaded back before final evaluation. Fixes the missing-checkpoint-restoration failure.

```
import pickle


class SimpleModel:
    def __init__(self):
        self.weight = [0.5, -0.25, 0.75]

    def state_dict(self):
        return {"weight": list(self.weight)}

    def load_state_dict(self, d):
        self.weight = list(d["weight"])


model = SimpleModel()
```

```
# The "best" checkpoint is saved to disk...
with open("best.pt", "wb") as fh:
    pickle.dump(model.state_dict(), fh)
```

```
# FIXED: restore the best checkpoint before final evaluation.
with open("best.pt", "rb") as fh:
    checkpoint = pickle.load(fh)
model.load_state_dict(checkpoint)
```
